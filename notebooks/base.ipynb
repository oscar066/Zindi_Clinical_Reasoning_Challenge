```python
from transformers import AutoTokenizer, DataCollatorForSeq2Seq

# Use the tokenizer for the model you plan to fine-tune
model_name = "t5-small"
tokenizer = AutoTokenizer.from_pretrained(model_name)
```

```python
# preprocess with the other columns as context
def preprocess_with_context(examples):
    # The target text
    targets = examples['clinician']
    
    # Build the structured input string for each example
    inputs = []
    for i in range(len(targets)):
        # Construct the context string
        context = f"context: county [{examples['county'][i]}] health_level [{examples['health_level'][i]}] experience [{examples['years_of_experience'][i]}]"
        
        # Combine with the prompt
        prompt = examples['prompt'][i]
        
        # Final input string
        input_string = f"{context} prompt: {prompt}"
        inputs.append(input_string)
        
    # Tokenize the structured inputs
    model_inputs = tokenizer(inputs, max_length=1024, truncation=True, padding="max_length")

    # Tokenize the targets
    with tokenizer.as_target_tokenizer():
        labels = tokenizer(targets, max_length=512, truncation=True, padding="max_length")
        
    model_inputs["labels"] = labels["input_ids"]
    return model_inputs
```

```python
# Apply the function to each dataset
tokenized_train_dataset = train_dataset["train"].map(preprocess_with_context, batched=True)

tokenized_test_dataset = train_dataset["test"].map(preprocess_with_context, batched=True)
```

```python
def compute_metrics(eval_preds):
    preds, labels = eval_preds
    
    # Handle the case where preds might contain invalid values
    try:
        # Clip predictions to valid token range
        vocab_size = tokenizer.vocab_size
        preds = np.clip(preds, 0, vocab_size - 1)
        
        # Replace -100 in labels with pad token
        labels = np.where(labels != -100, labels, tokenizer.pad_token_id)
        
        # Ensure all values are within valid range
        labels = np.clip(labels, 0, vocab_size - 1)
        
        # Convert to int32 to avoid overflow issues
        preds = preds.astype(np.int32)
        labels = labels.astype(np.int32)
        
        # Decode with error handling
        decoded_preds = []
        decoded_labels = []
        
        for pred_seq, label_seq in zip(preds, labels):
            try:
                decoded_pred = tokenizer.decode(pred_seq, skip_special_tokens=True)
                decoded_label = tokenizer.decode(label_seq, skip_special_tokens=True)
                decoded_preds.append(decoded_pred)
                decoded_labels.append(decoded_label)
            except Exception as e:
                print(f"Error decoding sequence: {e}")
                # Use empty strings as fallback
                decoded_preds.append("")
                decoded_labels.append("")
        
        # Process for ROUGE metric
        decoded_preds = ["\n".join(nltk.sent_tokenize(pred.strip())) if pred.strip() else "" 
                        for pred in decoded_preds]
        decoded_labels = ["\n".join(nltk.sent_tokenize(label.strip())) if label.strip() else "" 
                         for label in decoded_labels]
        
        # Compute ROUGE scores
        result = metric.compute(predictions=decoded_preds, references=decoded_labels, use_stemmer=True)
        return result
        
    except Exception as e:
        print(f"Error in compute_metrics: {e}")
        # Return default metrics if everything fails
        return {"rouge1": 0.0, "rouge2": 0.0, "rougeL": 0.0, "rougeLsum": 0.0}
```

```python
import torch

# 1. Test the trained model's generation
def test_model_generation(model, tokenizer, test_dataset, num_samples=5):
    """Test what the model is actually generating"""
    print("Testing model generation...")
    
    model.eval()
    
    for i in range(min(num_samples, len(test_dataset))):
        # Get a test sample
        sample = test_dataset[i]
        input_ids = torch.tensor([sample['input_ids']]).to(model.device)
        
        # Generate text
        with torch.no_grad():
            generated_ids = model.generate(
                input_ids=input_ids,
                max_length=128,
                num_beams=1,
                temperature=1.0,
                do_sample=False,
                pad_token_id=tokenizer.pad_token_id,
                eos_token_id=tokenizer.eos_token_id
            )
        
        # Decode input and output
        input_text = tokenizer.decode(sample['input_ids'], skip_special_tokens=True)
        generated_text = tokenizer.decode(generated_ids[0], skip_special_tokens=True)
        
        # Get the target text
        labels = [id if id != -100 else tokenizer.pad_token_id for id in sample['labels']]
        target_text = tokenizer.decode(labels, skip_special_tokens=True)
        
        print(f"\n--- Sample {i+1} ---")
        print(f"Input: {input_text[:200]}...")
        print(f"Generated: '{generated_text}'")
        print(f"Target: '{target_text}'")
        print(f"Generated length: {len(generated_text)}")
```

```python

# 2. Check if the model is generating anything meaningful
def analyze_generation_patterns(model, tokenizer, test_dataset, num_samples=10):
    """Analyze what patterns the model is generating"""
    print("Analyzing generation patterns...")
    
    generations = []
    model.eval()
    
    for i in range(min(num_samples, len(test_dataset))):
        sample = test_dataset[i]
        input_ids = torch.tensor([sample['input_ids']]).to(model.device)
        
        with torch.no_grad():
            generated_ids = model.generate(
                input_ids=input_ids,
                max_length=128,
                num_beams=1,
                do_sample=False,
                pad_token_id=tokenizer.pad_token_id,
                eos_token_id=tokenizer.eos_token_id
            )
        
        generated_text = tokenizer.decode(generated_ids[0], skip_special_tokens=True)
        generations.append(generated_text)
    
    # Analyze patterns
    empty_generations = sum(1 for gen in generations if len(gen.strip()) == 0)
    avg_length = sum(len(gen) for gen in generations) / len(generations)
    unique_generations = len(set(generations))
    
    print(f"Empty generations: {empty_generations}/{len(generations)}")
    print(f"Average length: {avg_length:.1f} characters")
    print(f"Unique generations: {unique_generations}/{len(generations)}")
    print(f"Sample generations: {generations[:3]}")
```

```python
# Run these tests
test_model_generation(model, tokenizer, tokenized_test_dataset)
analyze_generation_patterns(model, tokenizer, tokenized_test_dataset)
```

```python
# 1. Load fresh model larger model
print("Loading fresh T5-small model...")
model = AutoModelForSeq2SeqLM.from_pretrained("t5-base")
data_collator = DataCollatorForSeq2Seq(tokenizer=tokenizer, model=model)

# 2. Conservative training arguments
training_args = Seq2SeqTrainingArguments(
    output_dir="./clinical_t5_model_fixed_v2",
    eval_strategy="epoch",
    
    # Keep the working learning rate
    learning_rate=1e-6,
    
    # Batch settings
    per_device_train_batch_size=2,
    per_device_eval_batch_size=2,
    gradient_accumulation_steps=8,
    
    # Extend training
    num_train_epochs=4,  # More epochs to break the "True" pattern
    warmup_steps=20,
    weight_decay=0.01,
    
    # Better generation settings for evaluation
    predict_with_generate=True,
    generation_max_length=256,  # Allow longer generation
    generation_num_beams=4,     # More beams for better quality
    
    # Stability
    fp16=False,
    save_total_limit=3,
    logging_steps=5,
    save_strategy="epoch",
    max_grad_norm=1.0,
)


# 3. Updated compute_metrics with better error handling
def compute_metrics_conservative(eval_preds):
    preds, labels = eval_preds
    
    try:
        # Basic safety checks
        if preds is None or labels is None:
            return {"rouge1": 0.0, "rouge2": 0.0, "rougeL": 0.0, "rougeLsum": 0.0}
        
        # Handle predictions
        vocab_size = tokenizer.vocab_size
        preds = np.clip(preds, 0, vocab_size - 1).astype(np.int32)
        labels = np.where(labels != -100, labels, tokenizer.pad_token_id)
        labels = np.clip(labels, 0, vocab_size - 1).astype(np.int32)
        
        decoded_preds = []
        decoded_labels = []
        
        for pred_seq, label_seq in zip(preds, labels):
            try:
                # Decode predictions
                pred_text = tokenizer.decode(pred_seq, skip_special_tokens=True).strip()
                label_text = tokenizer.decode(label_seq, skip_special_tokens=True).strip()
                
                # Check for padding-only generation
                pred_tokens = [t for t in pred_seq if t != tokenizer.pad_token_id and t != tokenizer.eos_token_id]
                if len(pred_tokens) == 0:
                    pred_text = "model generated only padding tokens"
                
                # Ensure we have some text
                if len(pred_text) == 0:
                    pred_text = "no output generated"
                if len(label_text) == 0:
                    label_text = "no target text"
                
                decoded_preds.append(pred_text)
                decoded_labels.append(label_text)
                
            except Exception as e:
                print(f"Decode error: {e}")
                decoded_preds.append("decode error")
                decoded_labels.append("decode error")
        
        # Compute ROUGE
        result = metric.compute(
            predictions=decoded_preds, 
            references=decoded_labels, 
            use_stemmer=True
        )
        
        # Print sample for monitoring
        print(f"\nSample generation:")
        print(f"Pred: {decoded_preds[0][:100]}...")
        print(f"Target: {decoded_labels[0][:100]}...")
        
        return result
        
    except Exception as e:
        print(f"Metrics computation error: {e}")
        return {"rouge1": 0.0, "rouge2": 0.0, "rougeL": 0.0, "rougeLsum": 0.0}

# 4. Create new trainer with fresh model
trainer = Seq2SeqTrainer(
    model=model,  # Fresh model!
    args=training_args,
    train_dataset=tokenized_train_dataset,
    eval_dataset=tokenized_test_dataset,
    tokenizer=tokenizer,
    data_collator=data_collator,
    compute_metrics=compute_metrics_conservative,
)

print("Setup complete! Ready to train with conservative settings.")
print("Key changes:")
print("- Learning rate: 2e-5 → 1e-6 (20x lower)")
print("- Epochs: 5 → 2")
print("- Fresh model (not the broken one)")
print("- Better error handling")
print("- Gradient clipping")
print("\nRun: trainer.train()")
```

```python
# Test the current model's generation capabilities
def test_current_generation(model, tokenizer, test_dataset, num_samples=3):
    """Test the current model with different generation strategies"""
    
    model.eval()
    print("=== TESTING CURRENT MODEL GENERATION ===")
    
    for sample_idx in range(min(num_samples, len(test_dataset))):
        print(f"\n--- Sample {sample_idx + 1} ---")
        
        sample = test_dataset[sample_idx]
        input_ids = torch.tensor([sample['input_ids']]).to(model.device)
        
        # Get input text for context
        input_text = tokenizer.decode(sample['input_ids'], skip_special_tokens=True)
        target_text = tokenizer.decode([id if id != -100 else tokenizer.pad_token_id for id in sample['labels']], skip_special_tokens=True)
        
        print(f"Input: {input_text[:200]}...")
        print(f"Target: {target_text[:200]}...")
        
        # Try multiple generation strategies
        strategies = [
            {
                "name": "Greedy",
                "params": {
                    "max_new_tokens": 100,
                    "do_sample": False,
                    "num_beams": 1,
                    "pad_token_id": tokenizer.pad_token_id,
                    "eos_token_id": tokenizer.eos_token_id,
                }
            },
            {
                "name": "Beam Search", 
                "params": {
                    "max_new_tokens": 100,
                    "num_beams": 4,
                    "early_stopping": True,
                    "do_sample": False,
                    "pad_token_id": tokenizer.pad_token_id,
                    "eos_token_id": tokenizer.eos_token_id,
                }
            },
            {
                "name": "Sampling",
                "params": {
                    "max_new_tokens": 100,
                    "do_sample": True,
                    "temperature": 0.7,
                    "top_p": 0.9,
                    "pad_token_id": tokenizer.pad_token_id,
                    "eos_token_id": tokenizer.eos_token_id,
                }
            }
        ]
        
        for strategy in strategies:
            try:
                with torch.no_grad():
                    generated_ids = model.generate(
                        input_ids=input_ids,
                        **strategy['params']
                    )
                
                # Extract only new tokens
                input_length = input_ids.shape[1]
                generated_only = generated_ids[0][input_length:]
                generated_text = tokenizer.decode(generated_only, skip_special_tokens=True)
                
                print(f"{strategy['name']}: '{generated_text[:150]}{'...' if len(generated_text) > 150 else ''}'")
                
            except Exception as e:
                print(f"{strategy['name']}: Error - {e}")

# Test the current model
test_current_generation(model, tokenizer, tokenized_test_dataset)
```
